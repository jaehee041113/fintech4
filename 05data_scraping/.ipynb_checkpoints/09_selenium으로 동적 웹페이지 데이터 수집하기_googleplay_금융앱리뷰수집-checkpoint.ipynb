# selenium으로 동적 웹페이지 데이터 수집하기_googleplay_금융앱리뷰수집

### Selenium 사용하는 경우
- 데이터가 JavaScript 렌더링 이후에 생성되는 경우
- 스크롤/클릭 등 사용자 이벤트로 DOM이 바뀌는 경우

- BeautifulSoup/requests : 렌더링	HTML 원문(정적) 중심
- Selenium : 브라우저 렌더링(동적) 포함

### 구글 플레이는:
- 처음 페이지 열면 리뷰 몇 개만 로딩
- 스크롤해야 리뷰가 추가로 로딩됨
- 버튼 클릭(더보기)도 필요함
- -> 이건 HTML이 처음부터 다 안 내려옴 = 동적 웹페이지

- 그래서“사람처럼 브라우저를 조작하는 도구” = Selenium 을 씀

## 구글 플레이는 정적인 크롤링(requests)으로 안 되기 때문에 사람이 스크롤·클릭하는 것처럼 흉내 내서 금융 앱 리뷰를 자동으로 수집하는 코드

## 전체 흐름
```python
1. 크롬 브라우저를 자동으로 켠다
2. 구글 플레이 앱 페이지로 간다
3. 리뷰 창을 연다
4. 스크롤을 계속 내린다 (리뷰 더 불러오기)
5. 리뷰 하나하나를 찾아서
6. 별점 / 날짜 / 내용 뽑아서
7. 데이터프레임 → csv 저장
```

# selenium만 이용해서 데이터 추출하기

````python
# 실행하면 토스 웹페이지 열림

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By

# Chrome 옵션 설정
options = Options()
options.add_argument("--window-size=1280,900")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

# ChromeDriver 설정
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)
driver.set_page_load_timeout(30)

try:
    driver.get("https://play.google.com/store/apps/details?id=viva.republica.toss")
    print(driver.title)
finally:
    pass
````

````python
from selenium.webdriver.common.by import By

# 1개 요소 찾기 (WebElement 반환)
el = driver.find_element(By.CSS_SELECTOR, ".bARER")
print(el.get_attribute("outerHTML")) #태그내용 볼 때 .get_attribute("outerHTML") 써야 해서 불편함
print(el.text)

# 컨트롤+/ -> 전체주석처리
````

````python
# 여러 요소 찾기 (list[WebElement] 반환)

items = driver.find_elements(By.CSS_SELECTOR, ".EGFGHd")

# for item in items:
#     print(item.get_attribute("outerHTML"))  
````

````python
import pandas as pd
# 여러 요소 찾기 (list[WebElement] 반환)
items = driver.find_elements(By.CSS_SELECTOR, ".EGFGHd")
result = {}
for item in items:
#     print(item.text)
#     print(item.get_attribute('outerHTML'))
    name = item.find_element(By.CSS_SELECTOR, ".X5PpBb").text
    date = item.find_element(By.CSS_SELECTOR, ".bp9Aid").text
    rating = item.find_element(By.CSS_SELECTOR, ".iXRFPc").get_attribute('aria-label')
    rating = rating.split()[3][0]
    review_text = item.find_element(By.CSS_SELECTOR, ".h3YV2d").text
    keys = ("name", "date", "rating", "review_text")
    values = (name, date, rating, review_text)
    for key, value in zip(keys, values):
        result.setdefault(key, []).append(value)
df = pd.DataFrame(result)
df
````

# 토스 앱 리뷰하기 최종
### 사람이 크롬 켜서 → 토스 앱 들어가 → 리뷰창 열고 → 최신순 정렬 → 스크롤 내려가며 → 최근 28일 리뷰만 모아서 → DB에 저장하는 자동화 로봇
- 위의 내용 다 합침 (이 코드만 돌리면 됨)



- 셀레니움으로 토스 리뷰 페이지 접속- 자바스크립트로 평점리뷰 나올때까지 스크롤 내리기-열기-리뷰창열기 - 과거리뷰 로딩-마지막날짜 추출하기 -한번에 가져와서 반복문으로 리뷰 하나씩 꺼내서 result에 모으고 all_reviews에 모아서 데이터프레임 표로 만들기
- 주소만 바꾸면 다른 앱도 활용 가능

By.CSS_SELECTOR는
“HTML에서 특정 요소를 찾기 위한 주소 체계(찾는 방법)”야.

.bp9Aid는
👉 **CSS 선택자 문법으로 쓴 “주소”**고,
By.CSS_SELECTOR는
👉 “이 주소는 CSS 문법이야”라고 Selenium에게 알려주는 말이야.


“이건 CSS 선택자로 찾을 거야”
그래서 이렇게 씀: By.CSS_SELECTOR, ".bp9Aid"


📌 구조 : (어떤 방식으로 찾을지, 그 방식에 맞는 주소)

.bp9Aid ← 이건 진짜 CSS 맞아?
네가 한 말 100% 정확함 👇
“이거 f12 눌러서 html에서 클래스 가져온 거잖아?”
맞아.
HTML + CSS에서 쓰는 ‘CSS 선택자 문법’을 그대로 쓰는 거야.


CSS 선택자 기본 규칙 (딱 이것만 알면 됨)
문법	의미
.클래스명	class
#아이디	id
태그명	태그
태그.클래스	특정 클래스 가진 태그

**".bp9Aid" = class="bp9Aid" 인 요소**



````python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.keys import Keys
from datetime import datetime, timedelta
import pandas as pd
import time
from dbio import to_db

# Chrome 옵션 설정 
options = Options()
options.add_argument("--window-size=1280,900")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

# ChromeDriver 설정 - 크롬을 실제로 켜는 부분
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)
driver.set_page_load_timeout(30)

try: # driver = 나 대신 행동하는 손
    driver.get("https://play.google.com/store/apps/details?id=viva.republica.toss")
    print(driver.title)
finally:
    pass

wait = WebDriverWait(driver, 10)

# 자바스트립트로 윈도우를 0-1200px까지 스크롤 내리기
driver.execute_script("window.scrollTo(0, 1200)")
# 평점 및 리뷰 옆의 -> 버튼 찾아서 클릭하기
btn = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'button[aria-label="평점 및 리뷰 자세히 알아보기"]')))
btn.click()
# 최신순으로 리뷰를 정렬하기 위해 버튼 클릭
button = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, "#sortBy_1")))
button.click()
time.sleep(5)
# 최신을 찾아 클릭
button2 = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'span[aria-label="최신"]')))
button2.click()
time.sleep(5)


###########################################################
# 팝업된 리뷰창에서 데이터 수집하기
###########################################################

# 오늘 날짜를 추출하고 1달 전 날짜를 계산해 수집할 날짜 범위 정하기
# 날짜 기준 정하기 (28일) -> 오늘부터 28일 전까지만 수집하겠다
today = datetime.today()
end_date = today - timedelta(days=28)

# 리뷰 창을 조금씩 스크롤하면서‘최근 28일 이전 리뷰가 보일 때까지’ 계속 내리고,
# 그 다음에 화면에 보이는 모든 리뷰를 하나씩 뜯어서 날짜·별점·내용을 표로 만들어 DB에 저장한다

while True:
    
    # 리뷰창 스크롤 내려서 과거 리뷰 로딩하기
    # 리뷰 팝업 안에서만 스크롤
    #.fysCi.Vk3ZVd : 리뷰들이 들어있는 상자
    # scrollBy(0, 1000) : 아래로 1000px 내리기
    driver.execute_script("document.querySelector('.fysCi.Vk3ZVd').scrollBy(0, 1000)")
    time.sleep(1)
    
    # 리뷰 가장 마지막의 날짜 추출하기
    last_date = driver.find_elements(By.CSS_SELECTOR, ".bp9Aid")[-1].text
    last_date = last_date.replace("년 ", "-").replace("월 ", "-").replace("일", "")
    last_date = datetime.strptime(last_date, "%Y-%m-%d")
    n_reviews = len(driver.find_elements(By.CSS_SELECTOR, ".bp9Aid"))
    print("리뷰개수:", n_reviews, "end_date", end_date, "last_date", last_date, end="\r")
    
    # 언제 멈출지 판단하기 - 최근 28일보다 더 옛날 리뷰가 보이면 그만
    if last_date < end_date:
        break

all_reviews = []   # 진짜 수집 - 스크롤X, 이미 화면에 로딩된 리뷰들을 하나씩 분석
# 스크롤이 멈춘 후 데이터 수집하기
for item in driver.find_elements(By.CSS_SELECTOR, ".RHo1pe"):
    result = {} 
    # 리뷰에서 날짜 추출하기
    date = item.find_element(By.CSS_SELECTOR, ".bp9Aid").text
    date = date.replace("년 ", "-").replace("월 ", "-").replace("일", "")
    review_date = datetime.strptime(date, "%Y-%m-%d")
    # 평점 추출하기
    rating = item.find_element(By.CSS_SELECTOR, ".iXRFPc").get_attribute("aria-label")
    rating = rating.split()[3][0]
    # 리뷰 글 추출하기
    review_text = item.find_element(By.CSS_SELECTOR, ".h3YV2d").text
    
    # 딕셔너리로 묶기
    result["date"] = review_date
    result['rating'] = rating
    result['review_text'] = review_text
    all_reviews.append(result)
    
df = pd.DataFrame(all_reviews)
display(df) 
to_db("bank_app_reviews", "toss_reviews", df)
````
