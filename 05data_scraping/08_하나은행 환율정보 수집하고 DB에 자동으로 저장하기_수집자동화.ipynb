```python
# 하나은행 환율 수집 함수로 만들어주기
def exrate_get(ymd_dash, ymd):
    url="https://www.kebhana.com/cms/rate/wpfxd651_01i_01.do"
    payload=dict(ajax="true",tmpInqStrDt=ymd_dash,pbldDvCd=0,inqStrDt=ymd,
                 inqKindCd=1,requestTarget='searchContentDiv')
    r=requests.get(url, params=payload)
    df=pd.read_html(r.text)[0]
    df.insert(0,'날짜',ymd_dash)
    return df
```

# 멀티인덱스 컬럼명 단순화하기
- 위의 표 컬럼명 보기 불편함 -> 단순화해주기
- 경우 여러 개 있음
- 0 != 1 == 2 (예: '날짜'!= ''== '')
- 0 != 1 != 2
- 0 == 1 == 2

```python
# 컬럼명 출력 - 리스트 형태로 나옴
final_result.columns
```

```python
# 컬럼명 평탄화 함수
# 정보 딕셔너리로 정리 후, 함수로 만들기

def flatten_cols(df):
    
    new_cols={}
    for col in df.columns: # 컬럼 한 줄씩
        if col[0]!=col[1]==col[2]:
            if col[1]=="":
                new_cols[col]=col[0]
            else:
                new_cols[col]=col[0]+"_"+col[1].replace(" ","_")
        elif col[0]!=col[1]!=col[2]:
            new_cols[col]= " ".join(col).replace(" ","_")
        elif col[0]==col[1]==col[2]:
            new_cols[col]=col[0].replace(" ","_")

    return new_cols
```

# 날짜 데이터 생성 후 데이터 수집하고 column명 변환해서 DB에 저장하기

```python
from io import StringIO
import requests
import pandas as pd
import time
from datetime import datetime
from dbio import to_db
```

```python
# 하나은행 환율 수집 함수
def exrate_get(ymd_dash, ymd):
    url="https://www.kebhana.com/cms/rate/wpfxd651_01i_01.do"
    payload=dict(ajax="true",tmpInqStrDt=ymd_dash,pbldDvCd=0,inqStrDt=ymd,
                 inqKindCd=1,requestTarget='searchContentDiv')
    r=requests.get(url, params=payload)
    df=pd.read_html(StringIO(r.text))[0]
    df.insert(0,'날짜',ymd_dash)
    return df
```

```python
#컬럼명 평탄화 함수
def flatten_cols(df):
    
    new_cols={}
    for col in df.columns: # 컬럼 한 줄씩
        if col[0]!=col[1]==col[2]:
            if col[1]=="":
                new_cols[col]=col[0]
            else:
                new_cols[col]=col[0]+"_"+col[1].replace(" ","_")
        elif col[0]!=col[1]!=col[2]:
            new_cols[col]= " ".join(col).replace(" ","_")
        elif col[0]==col[1]==col[2]:
            new_cols[col]=col[0].replace(" ","_")

    return new_cols
```

```python
result=[]
for date in list(pd.date_range("2026-01-28", "2026-01-23", freq='-1B')): # 거꾸로 주는 옵션
    ymd_dash, ymd=date.strftime("%Y-%m-%d"), date.strftime("%Y%m%d")
    print(ymd_dash, ymd)
    data=exrate_get(ymd_dash, ymd)
    result.append(data)
final_result=pd.concat(result)

# 수집된 데이터프레임 컬럼명 평탄화 하기
new_cols_dict=flatten_cols(final_result)
new_cols= [new_cols_dict[col] for col in final_result]
final_result.columns=new_cols
final_result

# dbio 함수로 결과 DB에 저장하기
to_db("exchange_rage_data","exchange_rate", final_result)
```

# 오늘 날짜 환율 정보 수집하고 DB 저장하기

```python
from io import StringIO
import requests
import pandas as pd
import time
from datetime import datetime
from dbio import to_db
```

```python
# 하나은행 환율 수집 함수
def exrate_get(ymd_dash, ymd):
    url="https://www.kebhana.com/cms/rate/wpfxd651_01i_01.do"
    payload=dict(ajax="true",tmpInqStrDt=ymd_dash,pbldDvCd=0,inqStrDt=ymd,
                 inqKindCd=1,requestTarget='searchContentDiv')
    r=requests.get(url, params=payload)
    df=pd.read_html(StringIO(r.text))[0]
    df.insert(0,'날짜',ymd_dash)
    return df
```

```python
#컬럼명 평탄화 함수
def flatten_cols(df):
    
    new_cols={}
    for col in df.columns: # 컬럼 한 줄씩
        if col[0]!=col[1]==col[2]:
            if col[1]=="":
                new_cols[col]=col[0]
            else:
                new_cols[col]=col[0]+"_"+col[1].replace(" ","_")
        elif col[0]!=col[1]!=col[2]:
            new_cols[col]= " ".join(col).replace(" ","_")
        elif col[0]==col[1]==col[2]:
            new_cols[col]=col[0].replace(" ","_")

    return new_cols
```

# 오늘 날짜만 데이터 수집하기

```python
# 오늘 날짜 반환 함수
def t_day():
    today=datetime.today()
    date=today.date() # 날짜만 나옴
    ymd_dash, ymd = date.strftime("%Y-%m-%d"), date.strftime("%Y%m%d")
    return ymd_dash, ymd
```

```python
# t_day() 함수로 오늘 날짜 만든 후 데이터 수집

#data = exrate_get(*t_day()) # * : 내부 내용이 하나씩 풀어져서 들어감

ymd_dash, ymd=t_day()
data= exrate_get(ymd_dash, ymd)

# 수집된 데이터프레임 컬럼명 평탄화 하기
new_cols_dict=flatten_cols(data)
new_cols= [new_cols_dict[col] for col in data]
data.columns=new_cols

# dbio 함수로 결과 DB에 저장하기
to_db("exchange_rage_data","exchange_rate", data)

print(f"{ymd_dash} 환율 데이터 수집 및 저장 완료")
```
