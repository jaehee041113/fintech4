# matplotlib의 2가지 코딩 스타일

## 00-스타일 : matplotlib 공식 추천 스타일
- 복잡한 그래프를 그릴 때 사용
- 사용법이 pyplot스타일보다 복잡함

```python
x=np.linspace(0,2,100)
x
```

```python
fig, ax=plt.subplots(figsize=(5,2.7)) # fig의 사이즈 결정
print(fig)
print(ax) #그래프는 ax에 그림
ax.plot(x,x,label="linear") #선형
ax.plot(x,x**2, label='quadratic') #제곱
ax.plot(x,x**3, label='cubic') #세제곱
ax.set_xlabel('x_label')
ax.set_ylabel('y_label')
ax.set_title('간단 그래프')
ax.legend()
plt.show()

# 500x270 그래프 그림
```

```python
fig, ax=plt.subplots(3, 1, figsize=(5,12)) # 3행1열 그래프 나옴 -> 그래프3개 나옴
print(fig)
print(ax)

ax[0].plot(x,x,label="linear")
ax[1].plot(x,x**2, label='quadratic') #제곱
ax[2].plot(x,x**3, label='cubic') #세제곱

ax[0].set_xlabel('x_label')
ax[0].set_ylabel('y_label')
ax[0].set_title('간단 그래프')
ax[0].legend()

ax[1].set_xlabel('x_label')
ax[1].set_ylabel('y_label')
ax[1].set_title('quadratic 간단 그래프')
ax[1].legend()

ax[2].set_xlabel('x_label')
ax[2].set_ylabel('y_label')
ax[2].set_title('cubic 간단 그래프')
ax[2].legend()

plt.tight_layout() # 그래프 사이 간격 알아서 맞춰줌
plt.show()
```

# 선 모양 바꾸기 : set_linestyle(모양)

```python
data1, data2, data3, data4 = np.random.rand(4,100)
```

- 선 모양 : '-', '--', '-.',':',''

```python
fig,ax=plt.subplots(figsize=(5,2.7))
x=np.arange(len(data1))
ax.plot(x, np.cumsum(data1),color='blue', linewidth=3, linestyle=":") # : -> 점선으로 그려짐
ax.plot(x, np.cumsum(data1),color='orange', linewidth=2, linestyle="-.") # -. -> -.-...으로 그려짐

plt.show()
```

# 산점도 그래프 그리고 색 변경하기
- facecolor(안쪽색), edgecolor(테두리색)

```python
fig,ax=plt.subplots(figsize=(5,2.7))
ax.scatter(data1,data2, s=50, facecolor='C4', edgecolor='k')
plt.show()
```

- 컬러를 RGB값으로 주고 싶을 때 (0,1,0.5)처럼 0~1사이의 숫자를 R,G,B 순서의 튜플로 입력

```python
fig,ax=plt.subplots(figsize=(5,2.7))
ax.scatter(data1,data2, s=50, facecolor=(1,0.5,1), edgecolor=(1,0,0))
plt.show()
```
