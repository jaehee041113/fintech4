```python
# 반복문 안 하고 함수 짜기!!
def str2int(text):
    try:
        temp=int(text)
        return temp
    except:
        return 0
        
```

<div class="alert alert-success">
    <b>EXERCISE 42</b>: 생존율을 구하는 함수가 작성되어 있습니다. 성별 생존율을 groupby를 활용해 구하기 위해 ?부분을 알맞게 채워주세요.
</div>

```python
def survival_ratio(survived):
    return survived.sum() / len(survived)
```

```python
df[['Sex', "Survived"]].groupby('Sex').agg(survival_ratio)
# 기존 있는 함수인 .mean() 외에도 
# 직접 만든 함수를 agg에 넣어서 값 수정 후 출력해줄 수 있음
```
