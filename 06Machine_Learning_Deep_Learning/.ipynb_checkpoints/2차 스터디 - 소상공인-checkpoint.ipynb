1. 라이브러리 & 데이터 불러오기

```python
import pandas as pd
import numpy as np

# CSV 불러오기
store = pd.read_csv("./data/store_info.csv")
cash = pd.read_csv("./data/daily_cashflow_5000rows.csv")
fixed = pd.read_csv("./data/fixed_cost.csv")
loan = pd.read_csv("./data/loan_info.csv")
```

2. 데이터 전처리

```python
# 날짜 변환
cash['date'] = pd.to_datetime(cash['date'])

# 정렬
cash = cash.sort_values(['store_id','date'])
```

3. 파생 변수 생성

```python
# 7일 / 30일 평균 매출

cash['sales_7d_avg'] = cash.groupby('store_id')['total_sales'].transform(
    lambda x: x.rolling(7, min_periods=1).mean()
)

cash['sales_30d_avg'] = cash.groupby('store_id')['total_sales'].transform(
    lambda x: x.rolling(30, min_periods=1).mean()
)
```

```python
# 평균 지출
cash['total_expense'] = (
    cash['platform_fee'] +
    cash['ingredient_cost'] +
    cash['labor_cost'] +
    cash['utilities_cost'] +
    cash['marketing_cost'] +
    cash['other_expense'] +
    cash['loan_repayment']
)

cash['expense_30d_avg'] = cash.groupby('store_id')['total_expense'].transform(lambda x: x.rolling(30).mean())
```

```python
# 매출 추세 계수
cash['sales_trend'] = cash['sales_7d_avg'] / cash['sales_30d_avg']
```

4. 월말 잔액 예측

```python
# 남은 일수 가정 (예: 15일)
remaining_days = 15
```

```python
# 월말 잔액 공식
cash['predicted_end_balance'] = (
    cash['closing_balance']
    + (cash['sales_30d_avg'] * cash['sales_trend'] * remaining_days)
    - (cash['expense_30d_avg'] * remaining_days)
)
```

5. 단기 위험 계산

```python
# 고정비 합치기

fixed_cost = fixed[['store_id','fixed_cost_total']]

cash = cash.merge(fixed_cost, on='store_id', how='left')
```

```python
# 고정비 대비 잔액 위험
cash['fixed_cost_risk'] = (cash['fixed_cost_total'] / cash['closing_balance']) * 100
```

```python
# 매출 감소 위험
cash['sales_decline_rate'] = (cash['sales_30d_avg'] - cash['sales_7d_avg']) / cash['sales_30d_avg']

cash['sales_decline_risk'] = cash['sales_decline_rate'] * 100
```

```python
# 30일 내 마이너스 위험
cash['minus_risk'] = np.where(
    cash['predicted_end_balance'] < 0,
    90,
    20
)
```

```python
# 단기 위험 점수
cash['short_risk'] = (
    cash['fixed_cost_risk'] * 0.4
    + cash['sales_decline_risk'] * 0.3
    + cash['minus_risk'] * 0.3
)
```

6. 중기 위험 계산

```python
# 대출 정보 합치기
loan_info = loan[['store_id','monthly_payment']]

cash = cash.merge(loan_info, on='store_id', how='left')
```

```python
# 현금 감소 추세
cash['balance_90d'] = cash.groupby('store_id')['closing_balance'].shift(90)

cash['cash_decline_rate'] = (
    (cash['balance_90d'] - cash['closing_balance']) /
    (cash['balance_90d'] + 1)
)

cash['cash_decline_risk'] = cash['cash_decline_rate'].clip(lower=0) * 100
```

```python
# 고정비 증가율 (단순화)
cash['fixed_cost_growth'] = 0.05
cash['fixed_cost_growth_risk'] = cash['fixed_cost_growth'] * 100
```

```python
# 대출 부담
cash['loan_burden'] = cash['monthly_payment'] / cash['sales_30d_avg']
cash['loan_burden_risk'] = cash['loan_burden'] * 100
```

```python
# 중기 위험 점수
cash['mid_risk'] = (
    cash['cash_decline_risk'] * 0.5 +
    cash['loan_burden_risk'] * 0.3 +
    cash['fixed_cost_growth_risk'] * 0.2
)
```

7. 최종 위험 점수

```python
# 결측치 처리
# cash = cash.fillna(0)

cash['total_risk'] = (
    cash['short_risk'] * 0.6
    + cash['mid_risk'] * 0.4
)
```

8. 위험 등급

```python
def risk_level(score):
    if score < 30:
        return "안정"
    elif score < 60:
        return "주의"
    elif score < 80:
        return "위험"
    else:
        return "매우위험"

cash['risk_level'] = cash['total_risk'].apply(risk_level)
```

9. 최종 모델 출력

```python
cash['short_risk'] = cash['short_risk'].clip(0,100)
cash['mid_risk'] = cash['mid_risk'].clip(0,100)
cash['total_risk'] = cash['total_risk'].clip(0,100)

result = cash[[
'store_id',
'date',
'predicted_end_balance',
'short_risk',
'mid_risk',
'total_risk',
'risk_level'
]]

result.to_csv("risk_prediction_result.csv", index=False)

print(result.tail())
```

### 최종 모델이 출력하는 값

predicted_end_balance   → 월말 잔액 예측

short_risk              → 단기 위험

mid_risk                → 중기 위험

total_risk              → 종합 위험

risk_level              → 위험 등급
