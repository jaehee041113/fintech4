# 🌳 의사결정나무, 배깅, 랜덤포레스트, 부스팅의 차이

---

## 1️⃣ 의사결정나무 (Decision Tree)

의사결정나무는 **질문을 따라가며 답을 찾는 구조의 모델**이다.  
사람이 어떤 결정을 내릴 때처럼, “질문”을 하나씩 던지며  
조건에 따라 **결과를 단계적으로 선택하는 방식**이다.

```
예를 들어, “오늘 우산을 가져갈까?”라는 문제를 생각해 보자.

하늘에 구름이 많은가? → 예
↓

일기예보에서 비가 온다고 했는가? → 예
↓
👉 우산을 가져간다.
```

이처럼 **질문을 분기하며 조건에 따라 결론을 내리는 구조**가  
바로 의사결정나무이다.  
나무의 “가지(branch)”를 따라가다 보면 마지막 “잎(leaf)”에서 결론이 나온다.  

하지만 트리가 하나뿐이라,  
데이터가 조금만 달라져도 나무 모양이 크게 바뀌는 단점이 있다.  
즉, **데이터에 너무 민감하고 과적합(Overfitting)**이 잘 일어나는 모델이다.

👉 **요약**
- 장점: 이해하기 쉽고, 결정 이유를 설명하기 쉽다.  
- 단점: 데이터에 민감하고, 과적합이 잘 일어난다.  

---

## 2️⃣ 배깅 (Bagging)

배깅은 “Bootstrap Aggregating”의 줄임말이다.  
의사결정나무 한 그루로는 불안정하기 때문에,  
**여러 그루의 나무를 동시에 학습시켜 결과를 합치는 방법**이다.  

즉, 원래 데이터를 여러 번 섞어서 나눈 뒤  
각 나무가 조금씩 다른 데이터를 보고 학습하도록 한다.  
이렇게 학습한 나무들이 **투표(분류)**나 **평균(회귀)**으로 결과를 내린다.

예를 들어 “오늘 비가 올까?”라는 질문을  
세 그루의 나무에게 물어보면 다음과 같이 될 수 있다.
```
나무1: 비가 올 것이다 ☔
나무2: 비가 안 올 것이다 ☀️
나무3: 비가 올 것이다 ☔
→ 다수결 결과: 비가 올 것이다 ☔
```

이렇게 여러 나무가 함께 결정하기 때문에  
한 나무가 틀리더라도 전체 결과는 안정적이다.  
즉, **과적합이 줄고, 예측이 더 안정적**이 된다.  

배깅의 대표적인 알고리즘은 **랜덤 포레스트(Random Forest)**이다.  
이 방법은 “숲(forest)”처럼 여러 나무가 협력해서 예측을 하는 모델이다.

👉 **요약**
- 장점: 안정적이고, 과적합이 줄어든다.  
- 단점: 계산이 많고, 해석이 어렵다.  

---

## 3️⃣ 랜덤포레스트 (Random Forest)

랜덤포레스트는 **배깅(Bagging)**의 대표적인 알고리즘이다.  
즉, 여러 개의 의사결정나무를 만들어 **숲(forest)**처럼 사용하는 방법이다. 🌲🌲🌲  

배깅과 비슷하지만 한 가지 차이가 있다.  
랜덤포레스트는 **데이터뿐만 아니라, 사용할 특성(열)**도  
무작위로 일부만 선택해서 학습한다.  

예를 들어,  
나무마다 “하늘 상태”, “기온”, “습도”, “바람 세기” 중  
무작위로 몇 가지 정보만 보고 판단하도록 하는 것이다.  

이렇게 하면 각 나무가 **조금씩 다른 관점**에서 배우게 되어,  
서로 비슷하게 틀릴 가능성이 줄어든다.  
그 결과, 전체 숲의 예측이 더욱 안정적이 된다.  

👉 **요약**
- 배깅을 발전시킨 방법으로, 데이터와 특성을 모두 무작위로 사용한다.  
- 여러 나무가 협력하므로 예측이 안정적이고 강력하다.  
- 대표적인 앙상블(ensemble) 알고리즘이다.

👉 **비유**
- 한 나무(Decision Tree)는 **혼자 판단하는 친구**이다.  
- 여러 나무가 모인 랜덤포레스트는 **서로 다른 의견을 모아 다수결로 결정하는 반 친구들 모임**이다.  

---

## 4️⃣ 부스팅 (Boosting)

부스팅은 “여러 나무가 순서대로 배우며, 이전의 실수를 보완하는 방법”이다.  
즉, **앞의 모델이 틀린 부분을 다음 모델이 집중해서 학습**한다.  

예를 들어 공부할 때 틀린 문제를 복습하는 것과 같은 원리이다.  
첫 번째 모델이 만든 오차를 두 번째 모델이 고치고,  
두 번째 모델의 실수를 세 번째 모델이 보완하는 식이다.

이 과정을 반복하면 모델은 점점 더 똑똑해지고 정확해진다.  
하지만 너무 많이 반복하면, 훈련 데이터에만 맞춰 외우는 **과적합**이 생길 수 있다.  

대표적인 알고리즘에는 **XGBoost**, **LightGBM**, **CatBoost** 등이 있다.

👉 **요약**
- 장점: 학습을 반복할수록 예측력이 높아진다.  
- 단점: 너무 많이 학습하면 외우기만 해서 과적합이 생길 수 있다.  

---

## ⚖️ 5️⃣ 네 가지 방법의 비교표

| 구분 | 의사결정나무 | 배깅(Bagging) | 랜덤포레스트(Random Forest) | 부스팅(Boosting) |
|------|---------------|----------------|------------------------------|-------------------|
| 학습 방식 | 하나의 나무로 예측 | 여러 나무가 병렬로 학습 | 여러 나무가 병렬로 학습 (특성도 무작위) | 여러 나무가 순서대로 학습 |
| 학습 관계 | 독립적 | 독립적 | 독립적 | 서로 의존함 |
| 목표 | 단순한 규칙으로 예측 | 결과를 평균하여 안정화 | 무작위성으로 다양성 확보 + 안정화 | 이전 실수를 보완하며 정확도 향상 |
| 장점 | 이해 쉬움 | 과적합 완화, 안정적 | 성능 높고 안정적 | 예측력 매우 높음 |
| 단점 | 과적합 쉬움 | 느리고 복잡 | 계산량 많음 | 튜닝 어렵고 과적합 가능 |

---

## 🌟 6️⃣ 비유로 정리

- **의사결정나무**는 **혼자 판단하는 친구**이다.  
- **배깅**은 **서로 다른 친구들이 각자 생각한 뒤 투표하는 모둠**이다.  
- **랜덤포레스트**는 **서로 다른 시각(정보)으로 판단하는 친구들의 숲**이다.  
- **부스팅**은 **앞의 실수를 복습하면서 점점 똑똑해지는 학생**이다.  

---

## 🎯 7️⃣ 한 문장 요약

> 의사결정나무는 “하나의 나무로 판단하는 모델”이고,  
> 배깅은 “여러 나무가 함께 투표하는 방법”이며,  
> 랜덤포레스트는 “배깅을 발전시켜 더 다양한 시각에서 판단하는 숲”이고,  
> 부스팅은 “이전 실수를 점점 고쳐가며 발전하는 학습 방법”이다.

# 평가 지표 - 성능 지표 보는 거 중요함!! 보고 판단하기

# 데이터 분석 모델의 성능평가 - 중요!!

## 모델의 목적에 따라서 precision(정밀도)이 높은 것을 선택할지, recall(재현율/민감도)이 높은 것을 선택할 지 골라야 함

1. accuracy(정확도/정분류율): 1을 1이라고 맞추고 0을 0이라고 맞춘 비율
2. False Positive Rate(오분류율): 1을 0이라고 예측하고 0을 1이라고 예측해서 잘못 예측한 것
3. **precision(정밀도)**: 예측한 것 중에서 실제로 맞은 것, 예측 postsitve 중에서 실제 True인 비율
- -> 예측이 맞아야 하는 경우에 사용 - 수요 예측해서 상품 준비
4. **recall(재현율/민감도)**: 실제값을 실제값으로 잘 예측한 것, 실제 True중 예측 postive가 맞은 것
- -> 실제값이 중요한 경우 - 암진단
- precision(정밀도) 높으면/낮으면 -> recall(재현율/민감도) 낮음/높음
5. F1-Score: precision과 recall의 조화 평균, 
6. Specificity(특이도): 실제 nagative중 예측에 성공한 비율

# confusion matrix(혼동행렬) - 중요!! 개념, 표 알기
- 실제 : True/False
- 예측 : Positive/Nagative

-> 표에서 : 긍정/부정끼리 맞으면=True / 안 맞으면 = False

|              |  예측 Positive    | 예측 Nagative     |
|:--------------:|:------------------:|:-------------------:|
|**실제 True** | TP (True Positive) |FN (False Nagative)|
|**실제 False** | FP (False Positive)|TN (True Negative) |
* TP: 실제 Ture, 예측 Positive(실제 연봉 >5만, 예측 연봉 >5만)
* TN: 실제 False, 예측 Negative(실제 연봉 <=5만, 예측 연봉 <=5만)
* FP(1형 오류): 실제 True, 예측 Positive(실제 연봉 <=5만, 예측 연봉 >5만 )
* FN(2형 오류): 실제 True, 예측 Negative(실제 연봉 >5만, 예측 연봉 <=5만)

### 표 보고 공식 이해하기
* 정확도(accuracy_score): $ TP + TN \over TP+TN+FP+FN $
-> 긍정/부정끼리 맞는 정도
* 오분류율(FPR): $ FP+FN \over TP+TN+FP+FN $
* 정밀도(precision): $ TP \over TP+FP $ -> Positive라인 보기!(예측값 중요)
* 재현율/민감도(recall): $ TP \over TP+FN $ -> True라인 보기!(실제값 중요)
* 특이도(Specificity): $ TN \over FP+TN $
* F1-score: $  2 \cdot {Precision \cdot Recall \over Precision + Recall} $
