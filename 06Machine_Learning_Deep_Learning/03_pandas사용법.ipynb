# 데이터 프레임에서 일부 컬럼만 가져오기2 - 중요!!
### 앞쪽은 행을 슬라이싱, 뒤쪽은 열을 슬라이싱한 것!
- loc : 인덱스 이름과 컬럼명으로 데이터의 일부를 추출해서 가져옴 -> **이름**으로 가져옴
```python
df.loc[시작row인덱스 이름 : 끝row인덱스 이름, ['컬럼명1', '컬럼명2']]
```
- iloc : 인덱스 번호를 이용해서 슬라이싱으로 데이터의 일부를 가져옴 -> **인덱스**로 가져옴
```python
df.iloc[시작row인덱스 번호 : 끝row인덱스 번호+1, 시작column인덱스 번호 : 끝column인덱스 번호+1]
```

- loc(이름)이 덜 헷갈려서 평소에는 loc 더 많이 씀!

# 판다스 데이터 프레임에서 조건에 맞는 행만 가져오기
- 원하는 것만 가져옴
```python
df['컬럼명'] 조건식
```
- 조건식 연산자 >,<,>=,<=,==, !=, &, |


````python
# 승객번호가 3번인 행을 출력하세요
df[df['승객번호']==3]
# df['승객번호']==3 조건이 True인 것만 또 대괄호로 인덱스 함!
````

````python
# 나이가 35 초과인 사람만 필터링
df[df['나이']>35]
````

- 조건이 2개 이상인 경우 and 조건 : &
df[(df[컬럼명1]==조건1) & (df[컬럼명2] == 조건2)]

````python
# 성별이 여성이면서 나이가 35세 초과인 행찾기 & 조건
df[(df['성별'] == "female") & (df['나이'] > 35)]
````

````python
# 성별이 남성, 선실등급 1, 생존여부가 1인 사람들을 찾으세요
df[(df['성별'] == 'male') & (df['선실등급'] == 1) & (df['생존여부'] ==1)] 
````

````python
# 성별이 남성, 선실등급 1 혹은 2이고, 생존여부가 1인 사람들을 찾으세요
df[(df['성별'] == 'male') & ((df['선실등급'] == 1) | (df['선실등급'] == 2)) & (df['생존여부'] == 1)]
````

````python
# or = .isin 으로 처리 가능!
df[(df['성별'] == 'male') & (df['선실등급'].isin([1,2])) & (df['생존여부'] == 1)]
````
